## **Naturaleza del servicio (1p)**
### *Pérdidas*
Utilizando unos auriculares se simularán diferentes probabilidades de error en la recepción, utilizando un fragmento musical conocido a priori por el participante.

**Objetivo**: máxima probabilidad de error flujo A.

| Probabilidad | $25^{-4}$ a $75^{-2}$ | (6 puntos) |
|---|---|---|

```python
audio_file = 'res/Ode to Joy - Beethoven.wav'
```

Definición de la lista donde se encuentran los valores de la prueba. Cada uno de los valores representa la probabilidad de error por cada unidad de información que atraviese el sistema.

```python
import numpy as np
error_prob_set = np.linspace(25e-4, 75e-2, 6)

error_prob_set
```

```python
from lib.helper import play_audio_with_error
channel_map = (1, 1)

print("> NASVP-01: from 25^-4 to 75^-2 in 6 steps")

try:
    for ep in error_prob_set:
        print("\t- Playing audio with {:.3f} of error probability...".format(ep))
        play_audio_with_error(audio_file, channel_map, ep)

except KeyboardInterrupt:
    print("\n\nEarly stopping...")
```
