## **Naturaleza del servicio**
### *Pérdidas*
Utilizando unos auriculares se simularán diferentes probabilidades de error en la recepción, utilizando una grabación de una voz de una persona dando una secuencia de instrucciones.

**Objetivo**: máxima probabilidad de error flujo B.

| Probabilidad | $10^{-3}$ a $75^{-2}$ | (6 puntos) |
|---|---|---|

Texto enunciado en el audio:

```
Este es un fragmento preparado para una de las pruebas del trabajo fin de máster "Sistema para la comunicación síncrona punto a multipunto: aplicación a una orquesta sinfónica distribuida espacialmente".

La prueba consistirá en una secuencia de órdenes que se irán enunciando. Durante la prueba deberá prestar atención al nivel de claridad con el que es capaz de entender las órdenes.

El conjunto de órdenes se repetirá en seis ocasiones, empeorando su nivel de claridad perceptible de menos a más desfavorable. El objetivo consistirá en determinar el nivel máximo admisible.

- Asienta con la cabeza.
- Niegue con la cabeza.
- Levante la mano derecha.
- Levante la mano izquierda.
```

````python
from lib.helper import play_audio
INTRO = "res/Este es un fragmento - perdidas.wav"
play_audio(INTRO, (1, 1))
````

Definición de la lista donde se encuentran los valores de la prueba. Cada uno de los valores representa la probabilidad de error por cada unidad de información que atraviese el sistema.

````python
import numpy as np
error_prob_set = np.linspace(10e-3, 75e-2, 6)

error_prob_set
````

````python
from lib.helper import play_audio_with_error
channel_map = (1, 1)
AUDIO_FILE = "res/Instrucciones - {}.wav".format

print("> NASVP-02: from 10^-3 to 75^-2 in 6 steps")

try:
    for ep in error_prob_set:
        print("\t- Playing audio with {:.3f} of error probability...".format(ep))
        play_audio("res/beep.wav", channel_map)
        play_audio_with_error(AUDIO_FILE(np.random.randint(1,4)), channel_map, ep)

except KeyboardInterrupt:
    print("\n\nEarly stopping...")
````
