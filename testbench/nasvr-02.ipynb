## **Naturaleza del servicio (2p)**
### *Retraso de propagación*
Utilizando unos auriculares un participante recibirá una secuencia de órdenes improvisadas por otro participante que se encontrará cercano a este, pero no visible. En cada simulación se fijará un retardo estable y el participante que se encuentra lanzando las órdenes deberá evaluar el límite máximo para realizar una gestión eficiente.

**Objetivo**: máximo retardo perceptible.

| Retardo | 900-3000ms | (5 puntos) |
|---|---|---|

Definición de la lista donde se encuentran los valores de la prueba. Cada uno de los valores representa el retraso que puede experimentar cada unidad de información que atraviese el sistema.

Teniendo en cuenta que el retraso de la propia librería comprende los 400ms, no se ha considerado para el cálculo de cada uno de los puntos.

```python
import numpy as np

DELAY_OFFSET = 0.4 # pyadiuo delay offset in seconds

pre_delay_set = np.ndarray(shape=(5), dtype=float)
pre_delay_set = np.linspace(0.9, 3, 5)
delay_set = pre_delay_set - DELAY_OFFSET # restando el delay de la librería

delay_set
```

```python
from lib.helper import stream_audio

TESTBENCH_DURATION = 20 # seconds
print("> NASVP-02: from 900ms to 3000ms in 5 steps")

try:
    for delay in delay_set:
        print("\t- Streaming microphone with {:.1f}ms delay...".format(delay+DELAY_OFFSET), end="")
        stream_audio(TESTBENCH_DURATION, delay)
        
        print(" (press any key to continue)")
        input()

except KeyboardInterrupt:
    print("\n\nEarly stopping...")
```
