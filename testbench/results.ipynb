## **Resultados banco de pruebas**
### *Generación de gráficas*

Para poder adjuntar las gráficas resultantes del banco de pruebas se utilizan los siguientes fragmentos de código.

```python
import numpy as np
def map_answers(answer_list, value_list):
    return list(map(lambda x: value_list[x-1], answer_list))

import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-darkgrid')

def plot(unique_count_result, array, title, subtitle, xlabel, ylabel, xlim = (-.05,.8), width=0.05, unit=''):
    _, [ax, bx] = plt.subplots(1, 2, figsize=(8,4), width_ratios=[3,1])
    ax.bar(unique_count_result.values, unique_count_result.counts, width=width, align='center')

    ax.set_xticks(array)
    ax.set_xticklabels([f"{x:.3f}" for x in array], rotation=45)
    ax.set_xlim(xlim)
    ax.set_xlabel(xlabel)

    ax.set_yticks([0,1,2,3,4])
    ax.set_ylabel(ylabel)
    ax.set_title(subtitle)

    bp = bx.boxplot(unique_count_result.values, patch_artist=True, showfliers=True)
    bx.set_title('Intervalo de confianza')
    bx.set_xticks([])
    bx.set_xlabel('Q3 = {:.3f} {}'.format(np.percentile(unique_count_result.values, 75), unit))
    bp['boxes'][0].set(facecolor = 'wheat')

    plt.suptitle(title, fontweight="bold", fontsize=16, y=.95, x=.51)
    plt.tight_layout()
    plt.show()
```

### **NASVP-01** 

```python
error_prob_set_a = np.linspace(25e-4, 25e-2, 4)
error_prob_set_b = np.linspace(25e-2, 75e-2, 3)
error_prob_set = np.concatenate((error_prob_set_a[:-1], error_prob_set_b))
error_prob_set = np.array([0.0025, 0.085 , 0.1675, 0.25  , 0.5   , 0.75])
ANSWERS = [3,3,2,5]

nasvp1 = np.unique_counts(map_answers(ANSWERS, error_prob_set))
plot(
    nasvp1, error_prob_set,
     "NASVP-01", "Naturaleza del servicio (pérdidas)",
     "Probabilidad de error", "Respuestas"
)
nasvp1
```

### **NESCS-01** 

```python
delay_set = np.linspace(0.01, 0.5, 9)
ANSWERS = [3,4,5,3]

nescs1 = np.unique_counts(map_answers(ANSWERS, delay_set))
plot(
    nescs1, delay_set,
     "NESCS-01", "Nivel de sincronización (sincronización)",
     "Desfase en segundos", "Respuestas",
    xlim=(-.05,.55), width=0.04, unit='s'
)
nescs1
```
